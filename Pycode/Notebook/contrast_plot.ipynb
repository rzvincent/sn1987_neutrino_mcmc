```python
import corner
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import chi2

effective_data_file = [
    '15.01', '15.05', '16.43', '16.65', '16.99', '17.00', '17.07', '17.10', '17.40', 
    '17.48', '17.50', '17.51', '17.83', '18.04', '18.05', '18.09', '18.10', '18.50',
    '19.02', '19.56', '19.83', '19.99'
]


threshold = chi2.ppf(0.95, df=2) # 95% confidence level for 2 parameters

fig, ax = plt.subplots()

arr_within_2d = []
arr_outside_2d = []

# extract 2d data
for file_id in effective_data_file:
    path_2d = f'./Bin/smellycat260817/data/InverseS/2d_m{file_id}_5e5_p32_IS_1e-2Tnu.npy'
    arr_2d = np.load(path_2d) # shape (3200, 2)
    
    mean_2d = np.mean(arr_2d, axis=0)
    cov_2d = np.cov(arr_2d, rowvar=False)
    inv_cov_2d = np.linalg.inv(cov_2d)
    diff_2d = arr_2d - mean_2d
    mahalanobis_sq = np.sum(np.dot(diff_2d, inv_cov_2d) * diff_2d, axis=1)

    # Data points WITHIN 95% confidence
    within_threshold_2d = mahalanobis_sq <= threshold

    arr_within_2d.append(arr_2d[within_threshold_2d])
    arr_outside_2d.append(arr_2d[~within_threshold_2d])

arr_within_2d = np.vstack(arr_within_2d)
arr_outside_2d = np.vstack(arr_outside_2d)

ax.scatter(10**arr_within_2d[:, 0], 10**arr_within_2d[:, 1], s=10, alpha=0.05, color='orange')

ax.tick_params(axis='both', which='both', direction='in', top=True, right=True, labelsize=12)
ax.set_title('2D model samples with 95% confidence region')
ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlabel(r'$m_{\phi}$ (eV)')
ax.set_ylabel(r'$\lambda_{\phi\nu}$')
ax.set_xlim(1, 1e4)
ax.set_ylim(1e-9, 1e-4)

fig
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binned_statistic
from scipy.signal import savgol_filter  # Import the smoothing filter


color_1 = '#f76f81'  # A shade of red/pink for osc
color_2 = '#006FED'  # A shade of blue
color_3 = '#d743d7'  # A shade of purple

# Hyper-K projection data

m_space = np.logspace(0, 4, 1000)  # eV
m_x, l_y = 10**(np.load(r'.\Data\m_lambda_space_2d.npy').T)
m_x_osc, l_y_osc = 10**(np.load(r'.\Data\m_lambda_space_2d_osc.npy').T)
m_x_tpi, l_y_tpi = 10**(np.load(r'.\Data\m_lambda_space_2d_osc_TPI.npy').T)

def clean_array(y_data):
    mask = np.isfinite(y_data)
    if not mask.all():
        # Interpolate the bad values using the good values
        y_data = np.interp(np.arange(len(y_data)), np.arange(len(y_data))[mask], y_data[mask])
    return y_data

l_y_osc = clean_array(l_y_osc)
l_y_tpi = clean_array(l_y_tpi)

from scipy.signal import savgol_filter # apply if you want a smoothing filter
window_size = 50  # Choose an odd window size
l_y_osc = savgol_filter(l_y_osc, window_length=window_size, polyorder=3)  
l_y_tpi = savgol_filter(l_y_tpi, window_length=window_size, polyorder=3)  

# 1d and 2d data within 95% confidence region
x_2d = arr_within_2d[:, 0]
y_2d = arr_within_2d[:, 1]

x_2d_osc = arr_within_2d_osc[:, 0]
y_2d_osc = arr_within_2d_osc[:, 1]

x_2d_tpi = arr_within_2d_tpi[:, 0]
y_2d_tpi = arr_within_2d_tpi[:, 1]

# the number of bins for the binned statistic
num_bins = 200

y_max_vals_2d, bin_edges_2d, _ = binned_statistic(x_2d, y_2d, statistic='max', bins=num_bins)
y_max_vals_2d_osc, bin_edges_2d_osc, _ = binned_statistic(x_2d_osc, y_2d_osc, statistic='max', bins=num_bins)
y_max_vals_2d_tpi, bin_edges_2d_tpi, _ = binned_statistic(x_2d_tpi, y_2d_tpi, statistic='max', bins=num_bins)

x_centers_2d = (bin_edges_2d[:-1] + bin_edges_2d[1:]) / 2
x_centers_2d_osc = (bin_edges_2d_osc[:-1] + bin_edges_2d_osc[1:]) / 2
x_centers_2d_tpi = (bin_edges_2d_tpi[:-1] + bin_edges_2d_tpi[1:]) / 2

# smooth if needed
y_max_vals_2d = savgol_filter(y_max_vals_2d, window_length=30, polyorder=3)
y_max_vals_2d_osc = savgol_filter(y_max_vals_2d_osc, window_length=30, polyorder=3)
y_max_vals_2d_tpi = savgol_filter(y_max_vals_2d_tpi, window_length=30, polyorder=3)

plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman']
plt.rcParams['mathtext.fontset'] = 'stix' 
# STIX fonts look very similar to Times New Roman

fig, ax = plt.subplots(figsize=(8, 6))

label_fontsize = 22
tick_fontsize = 18

# ax.scatter(x, y, s=10, alpha=0.1, color='steelblue', label='Data')
ax.tick_params(axis='both', which='both', direction='in', 
               top=True, right=True, labelsize=12)


# Plot the extracted upper limit line
ax.plot(10**x_centers_2d, 10**y_max_vals_2d, linewidth=2, color = color_2)
ax.plot(10**x_centers_2d_osc, 10**y_max_vals_2d_osc, linewidth=2, color = color_1)
ax.plot(10**x_centers_2d_tpi, 10**y_max_vals_2d_tpi, linewidth=2, color = color_3)

ax.plot(m_x, l_y, color=color_2, linestyle='--', zorder=15, linewidth=2.2)
ax.plot(m_x_osc, l_y_osc, color=color_1, linestyle='--', zorder=15, linewidth=2.2)
ax.plot(m_x_tpi, l_y_tpi, color=color_3, linestyle='--', zorder=15, linewidth=2.2)

# Raffelt SN1987A limit
m_x = m_space
g_y = 6.06e-4/m_x # from 3.71e-4 to 6.06e-4
ax.plot(m_x, g_y, color='black', linestyle=':', alpha = 0.5)

# CMB data

def get_cmb_neff_extended(m_phi):
    """
    Returns the smoothed g_nu upper bound for CMB N_eff.
    """
    y0 = 1.6398e-5  
    xc = 23.0021    
    b = -1.0766     
    p = 2.7831      
    
    return y0 * (1.0 + (m_phi / xc)**p)**(b / p)

m_cmb = m_space
g_cmb = get_cmb_neff_extended(m_cmb)

# m_cmb, g_cmb = np.loadtxt(r'F:\Neutrino_SI\Bin\dataset_xu_cmb_neff.csv', delimiter=',', unpack=True)

# print(m_cvb, g_cvb)
ax.plot(m_cmb, g_cmb, color='black', alpha = 0.5)
ax.fill_between(m_cmb, g_cmb, 1e-2, color='black', alpha=0.1)

# 0vββ data
ax.axhline(1e-5, color='black', alpha = 0.3, linestyle='--')
# ax.fill_between(m_x, 1e-5, 1e-4, color='purple', alpha=0.1)

ax.set_xscale('log')
ax.set_yscale('log')
ax.set_xlabel(r'$m_{\phi}$ (eV)', fontsize=label_fontsize)
ax.set_ylabel(r'$\lambda_{\phi\nu}$', fontsize=label_fontsize)
ax.tick_params(axis='both', which='both', direction='in', top=True, right=True, labelsize=tick_fontsize)
ax.set_xlim(1, 1e4)
ax.set_ylim(5e-12, 5e-4)
# remove legend frame 

# Frame parameters for the text and lines
x_shift, y_shift = 0.2, 0.06
# row 1: Flavor blind and w. Oscillation and tpi
x_fb = 0.3
x_wo = x_fb+x_shift
x_tpi = x_wo+x_shift
y1 = 0.2

# column 1: SN1987A and Hyper-K Proj.
x_sn = x_fb - 0.17
x_hk = x_sn
y_sn = y1 - y_shift
y_hk = y_sn - y_shift

half_length = 0.04
text_fontsize = 16

ax.text(x_fb, y1, 'Flavor Blind', transform=ax.transAxes, fontsize=text_fontsize, color='k', ha='center', va='center')
ax.text(x_wo, y1, r'w. Osc', transform=ax.transAxes, fontsize=text_fontsize, color='k', ha='center', va='center')
ax.text(x_tpi, y1, r'w. Osc', transform=ax.transAxes, fontsize=text_fontsize, color='k', ha='center', va='center')
ax.text(x_wo, y1+y_shift/1.5, r'$(\lambda_{\phi\nu} \propto \mathrm{const})$', transform=ax.transAxes, fontsize=text_fontsize, color='k', ha='center', va='center')
ax.text(x_tpi, y1+y_shift/1.5, r'$(\lambda_{\phi\nu} \propto m_1)$', transform=ax.transAxes, fontsize=text_fontsize, color='k', ha='center', va='center')

ax.text(x_sn, y_sn, 'SN1987A', transform=ax.transAxes, fontsize=text_fontsize, color='k', ha='center', va='bottom')
ax.text(x_hk, y_hk, 'Hyper-K Proj.', transform=ax.transAxes, fontsize=text_fontsize, color='k', ha='center', va='bottom')

ax.plot([x_fb-half_length, x_fb+half_length], [y_sn + y_shift/4, y_sn + y_shift/4], color=color_2, transform=ax.transAxes) 
ax.plot([x_fb-half_length, x_fb+half_length], [y_hk + y_shift/4, y_hk + y_shift/4], color=color_2, linestyle='--',transform=ax.transAxes) 
ax.plot([x_wo-half_length, x_wo+half_length], [y_sn + y_shift/4, y_sn + y_shift/4], color=color_1, transform=ax.transAxes) 
ax.plot([x_wo-half_length, x_wo+half_length], [y_hk + y_shift/4, y_hk + y_shift/4], color=color_1, linestyle='--',transform=ax.transAxes) 
ax.plot([x_tpi-half_length, x_tpi+half_length], [y_sn + y_shift/4, y_sn + y_shift/4], color=color_3, transform=ax.transAxes) 
ax.plot([x_tpi-half_length, x_tpi+half_length], [y_hk + y_shift/4, y_hk + y_shift/4], color=color_3, linestyle='--',transform=ax.transAxes) 


# ax.plot([x_3, x_4], [y_1, y_1], color=color_2, transform=ax.transAxes, linestyle='--', linewidth=2.5)
# ax.plot([x_3, x_4], [y_2 , y_2 ], color=color_1, transform=ax.transAxes, linestyle='--', linewidth=2.5)


ax.text(0.70, 0.54, r'$\mathrm{CMB} ~ N_{\mathrm{eff}}$', transform=ax.transAxes, fontsize=text_fontsize, color='grey', ha='center', va='bottom', rotation=-22, zorder=10)
ax.text(0.44, 0.73, r'Fiorillo et al.(2023)', transform=ax.transAxes, fontsize=text_fontsize, color='grey', ha='center', va='bottom', rotation=-21, zorder=10)
ax.text(0.15, 0.74, r'$0\nu\beta\beta\phi$', transform=ax.transAxes, fontsize=text_fontsize, color='grey', ha='center', va='bottom', zorder=10)
# ax.legend(frameon=False, fontsize=12)

plt.tight_layout()
plt.savefig(r".\Plots\contrast_plot.pdf", bbox_inches='tight')
fig
```
